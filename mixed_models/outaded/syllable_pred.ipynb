```python
# Mapping from float code to 3-digit string
mapping = {0.0:'000', 1.0:'100', 2.0:'200', 3.0:'300', 4.0:'400', 5.0:'500', 6.0:'600', 7.0:'700',
           8.0:'010', 9.0:'110', 10.0:'210', 11.0:'310', 12.0:'410', 13.0:'510', 14.0:'610', 15.0:'710',
           16.0:'001', 17.0:'101', 18.0:'201', 19.0:'301', 20.0:'401', 21.0:'501', 22.0:'601', 23.0:'701',
           24.0:'011', 25.0:'111', 26.0:'211', 27.0:'311', 28.0:'411', 29.0:'511', 30.0:'611', 31.0:'711'}

def onehot(codes, mapping):
    # Handle NaNs separately
    valid_mask = ~np.isnan(codes)
    codes_valid = codes[valid_mask]

    # Vectorized: convert float codes to strings
    codes_str = np.array([mapping[c] for c in codes_valid])
    # Split strings into characters -> 2D array
    codes_split = np.array([list(s) for s in codes_str], dtype=int)  # shape (n_valid, 3)

    # Initialize arrays
    n_trials = len(codes)
    paw = np.full(n_trials, np.nan)
    whisk = np.full(n_trials, np.nan)
    lick = np.full(n_trials, np.nan)

    paw[valid_mask] = codes_split[:,0]
    whisk[valid_mask] = codes_split[:,1]
    lick[valid_mask] = codes_split[:,2]
    
    # --- One-hot encode paw (8 states) ---
    n_paw = 8
    paw_ohe = np.zeros((n_trials, n_paw), dtype=int)
    paw_ohe[valid_mask, paw[valid_mask].astype(int)] = 1

    # One-hot encode whisk and lick
    whisk_col = np.full(n_trials, -1)
    lick_col = np.full(n_trials, -1)
    whisk_col[valid_mask] = whisk[valid_mask]
    lick_col[valid_mask] = lick[valid_mask]

    # Concatenate all one-hot features into a single matrix
    X_features = np.concatenate([paw_ohe, whisk_col[:, None], lick_col[:, None]], axis=1)
    
    return X_features
```

```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
# data = pd.DataFrame(simplex)
data = pd.DataFrame(np.array(session_syllables))

data['mouse_name'] = session_syllables.reset_index()['mouse_name'].copy()
data['session'] = session_syllables.reset_index()['session'].copy()
data = data.rename(columns={col:str(col) for col in data.columns})

features = [col for col in data.columns if col not in ['mouse_name', 'session']]
group_col = 'mouse_name'

random_effects = pd.DataFrame(columns=features)
ICC = []

for feature in features:
    
    # Basic checks
    if feature not in data.columns:
        print(f"Skipping {feature}: not in df")
        continue

    # Convert to numeric (coerce non-numeric -> NaN)
    data[feature] = pd.to_numeric(data[feature], errors='coerce')

    sub = data.rename(columns={feature:'feature'}).copy()
    sub = sub[['mouse_name', 'feature', 'session']]
    # sub['feature'] = np.sqrt(sub['feature'])
    # Test with shuffle
    # sub['feature'] = np.random.permutation(np.sqrt(sub['feature']))
    formula = f"feature ~ 1"

    model = smf.mixedlm(formula, sub, groups=sub[group_col])
    res = model.fit(reml=True, method='lbfgs', warn_convergence=False)

    if res.cov_re.iloc[0,0] > 0:
        # print(res.summary())
        # Extract random intercepts: result.random_effects is dict mouse -> {Group: value} (naming depends)
        re = res.random_effects

        # Convert to Series with mouse index -> intercept
        # The dictionary format can vary; inspect `re` if needed
        vals = {}
        for mid, rdict in re.items():
            # rdict is usually {'Group': intercept} or {'Intercept': value}
            # pick the first numeric value
            if isinstance(rdict, dict):
                # find first numeric value
                num = None
                for k, v in rdict.items():
                    if np.isscalar(v):
                        num = float(v)
                        break
                vals[mid] = num
            else:
                # sometimes it's already a scalar
                vals[mid] = float(rdict)

        random_effects[feature] = pd.Series(vals, name=feature)

        icc = res.cov_re.iloc[0,0] / (res.cov_re.iloc[0,0] + res.scale)
        if len(ICC) == 0:
            ICC = [icc]
        else:
            ICC = np.concatenate([ICC, [icc]])
    else:
        icc = res.cov_re.iloc[0,0] / (res.cov_re.iloc[0,0] + res.scale)
        ICC = np.concatenate([ICC, [icc]])
```
